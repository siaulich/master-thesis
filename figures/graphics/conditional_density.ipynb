```python
import matplotlib.pyplot as plt
import numpy as np
import scipy.stats as stats
import atlas_mpl_style as mpl
mpl.use_atlas_style()
plt.rcParams["font.size"] = 20

```

```python
gaussian_left = stats.norm(loc=-2, scale=0.5)
gaussian_right = stats.norm(loc=2, scale=0.5)

x = np.linspace(-4, 4, 1000)

fig, ax = plt.subplots(figsize=(10, 6))
ax.fill(x, gaussian_left.pdf(x) + gaussian_right.pdf(x), color='tab:blue')
ax.axvline(0, color='tab:orange', linestyle='--', linewidth=3)
ax.set_xlim(-4, 4)
ax.set_ylim(0, 1)
ax.set_xticks([0])
ax.set_xticklabels([r"$\mathbb{E}_{(\mathbf{y}|\mathbf{x})}[\mathbf{y}]$"])
ax.get_yaxis().set_visible(False)
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)
ax.spines['left'].set_visible(False)
ax.text(-3.1, 0.8, r'$p(\mathbf{y}|\mathbf{x})$')
fig.tight_layout()
fig.savefig("multimodal_probability.pdf", bbox_inches='tight', dpi=300)
```
